# Step 1: A single qubit

A qubit's state is a unit vector in $\mathbb{C}^2$:

$$|\psi\rangle = \alpha|0\rangle + \beta|1\rangle, \qquad |\alpha|^2 + |\beta|^2 = 1$$

where $|0\rangle = \begin{pmatrix}1\\0\end{pmatrix}$ and $|1\rangle = \begin{pmatrix}0\\1\end{pmatrix}$.

Gates are 2×2 **unitary** matrices: $U^\dagger U = I$. Applying a gate is matrix–vector multiplication.

```python
%load_ext autoreload
%autoreload 2

import numpy as np

from qsim import gates
from qsim.state import apply, ket0, ket1
```
